POWER TRANSFORMATION

```python
from sklearn.preprocessing import PowerTransformer
```

```python
X= np.array([
    [-1],
    [0],
    [3],
    [5],
    [10],
    [50] 
])
pt = PowerTransformer(method='yeo-johnson')
X_transformed = pt.fit_transform(X)
print('original:')
print(X.ravel())

print('\nTransformed:')
print(np.round(X_transformed.ravel(),3))
print('\nLearned lambda:')
print(pt.lambdas_)
```

```python
y= np.array([
    [1],
    [2],
    [3],
    [5],
    [10],
    [50] 
])
pt = PowerTransformer(method='box-cox')
y_transformed = pt.fit_transform(y)
print('original:')
print(y.ravel())

print('\nTransformed:')
print(np.round(y_transformed.ravel(),3))
print('\nLearned lambda:')
print(pt.lambdas_)
```

```python
import numpy as np
import pandas as pd

# Creating a sample dataset with missing values in numerical and categorical columns
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    # Numerical columns (with nulls)
    'Age': [25, np.nan, 30, 35, np.nan, 28, 42, 39],
    'Salary': [50000, 64000, np.nan, 80000, 52000, np.nan, 95000, 72000],
    'Experience_Years': [2, 5, 7, np.nan, 1, 4, 15, 11],
    # Categorical columns (with nulls)
    'Department': ['IT', 'HR', np.nan, 'Finance', 'IT', 'Marketing', np.nan, 'Finance'],
    'City': ['New York', np.nan, 'Chicago', 'Austin', 'New York', np.nan, 'Chicago', 'Austin'],
    'Gender': ['Male', 'Female', 'Female', np.nan, 'Male', 'Female', 'Male', np.nan],
}

df = pd.DataFrame(data)

# Check null values count and data types
print("--- Missing Values Count ---")
print(df.isnull().sum())
print("\n--- Data Sample ---")
display(df)
```

```python
from sklearn.impute import SimpleImputer
```

```python
impute = SimpleImputer(strategy='mean')
impute.fit_transform(df[['Salary']]).flatten()
```

```python
impute = SimpleImputer(strategy='mean')
df['new_salary']= impute.fit_transform(df[['Salary']]).flatten()
```
